```python
# Fixed version of the first cell: re-define configuration and the API call helper.
# (Relies on `requests` and `json` already imported in a previous cell.)

API_BASE_URL = "https://api-ap-southeast-1.modelarts-maas.com"
API_PATH = "/v1/chat/completions"
API_URL = f"{API_BASE_URL}{API_PATH}"
AUTH_TOKEN = "vTq-UE2K7xcrt08Kn5LPQeTNfsYkA3CqAEKnK_wkhePATbk87UQY_byvmAHrfYKLGYefWvUUWoYms053d-xlog"  # Keep this secret!

headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {AUTH_TOKEN}"
}

def call_competition_model(model_name, user_prompt, system_prompt=None, **generation_params):
    """Call the competition model API and return generated text or None on failure."""
    messages = []
    if system_prompt:
        messages.append({"role": "system", "content": system_prompt})
    messages.append({"role": "user", "content": user_prompt})

    payload = {
        "model": model_name,
        "messages": messages,
        **generation_params
    }

    try:
        print(f"Sending request to {API_URL} (model={model_name})...")
        response = requests.post(API_URL, headers=headers, json=payload, timeout=60)
        response.raise_for_status()
        response_data = response.json()
    except requests.exceptions.RequestException as e:
        print(f"API request failed: {e}")
        if hasattr(e, "response") and e.response is not None:
            try:
                print(f"Status Code: {e.response.status_code}")
                print(f"Error Response: {e.response.text}")
            except Exception:
                pass
        return None

    try:
        # Expected structure: {'choices': [{'message': {'content': '...'}}]}
        generated_text = response_data["choices"][0]["message"]["content"]
        return generated_text.strip()
    except (KeyError, IndexError, TypeError) as e:
        print(f"Failed to parse API response: {e}")
        print("Response JSON:", json.dumps(response_data, indent=2))
        return None
```

```python
# Example usage (won't work without a valid token):
answer = call_competition_model("deepseek-v3.1", "กรุงเทพมีกี่เขตและแต่ละเขตมีกี่แขวง", max_tokens=10000, temperature=0.1)
print(answer)
```
